# Limpeza do Dataset




```python
# Importando Bibliotecas
import pandas as pd
import numpy as np
```

```python
# Carregando o Dataset
vendas_cafe1 = pd.read_csv("dirty_cafe_sales.csv")
```

```python
vendas_cafe = vendas_cafe1.copy()
```

```python
# Dados sujos
vendas_cafe.head(-5)
```

```python
# Juntar os nomes das colunas com ifen_
vendas_cafe = vendas_cafe.rename(columns={
    'Transaction ID': 'Transaction_ID',
    'Price Per Unit':'Price_Per_Unit',
    'Total Spent': 'Total_Spent',
    'Payment Method': 'Payment_Method',
    'Transaction Date': 'Transaction_Date'
})
```

```python
# Informacoes basicas do Dataset
vendas_cafe.info()
```

```python
#Verificando valores Unicos por colunas e identificando valores incorretos
print(vendas_cafe['Item'].value_counts())
print(vendas_cafe['Quantity'].value_counts())
print(vendas_cafe['Price_Per_Unit'].value_counts())

print(vendas_cafe['Payment_Method'].value_counts())
print(vendas_cafe['Total_Spent'].value_counts())
```

```python
# Verificando valores incorretos, porcentegem que representa ao dataset
faltantes_pct = (vendas_cafe.isnull().sum() / len(vendas_cafe)) *100
print(vendas_cafe.isnull().sum())
print('\nPorcentagem de dados incorretos por coluna:')
print(f'{faltantes_pct.sort_values(ascending=False)}')
```

```python
# Substituindo os valores sujos por NaN usando o numpy
#Criado uma lista de valores alem do que foi encontrado
dados_sujo = ['ERROR', 'UNKNOWN', 'None', 'N/A', 'NA', '-', 'null', '']
vendas_cafe = vendas_cafe.replace(dados_sujo, np.nan)
```

```python
vendas_cafe.head(-5)
```

```python
# verificando valores nulos
vendas_cafe.isnull().sum()

```

```python
#Verificando valores unicos das colunas categoricas
print(vendas_cafe['Item'].unique())
print(vendas_cafe['Quantity'].unique())
print(vendas_cafe['Payment_Method'].unique())
print(vendas_cafe['Location'].unique())
```

```python
# Mudar os tipos das colunas
vendas_cafe['Item'] = vendas_cafe['Item'].astype('category')
vendas_cafe['Quantity'] = pd.to_numeric(vendas_cafe['Quantity'], errors='coerce').astype('Int64')
vendas_cafe['Price_Per_Unit'] = pd.to_numeric(vendas_cafe['Price_Per_Unit'], errors='coerce').astype('float64')
vendas_cafe['Total_Spent'] = pd.to_numeric(vendas_cafe['Total_Spent'], errors='coerce').astype('float64')
vendas_cafe['Payment_Method'] = vendas_cafe['Payment_Method'].astype('category')
vendas_cafe['Location'] = vendas_cafe['Location'].astype('category')
vendas_cafe['Transaction_Date'] = pd.to_datetime(vendas_cafe['Transaction_Date'], errors='coerce')
```

```python
vendas_cafe.info()
```

```python
# Separar Dia da Semana, Mês e Ano
vendas_cafe['Week_Day'] = vendas_cafe['Transaction_Date'].dt.day_name()
vendas_cafe['Day'] = vendas_cafe['Transaction_Date'].dt.day
vendas_cafe['Month'] = vendas_cafe['Transaction_Date'].dt.month
vendas_cafe['Year'] = vendas_cafe['Transaction_Date'].dt.year
```

```python
vendas_cafe.head()
```

```python
# Mudar os tipos de coluna Day_week para cateria,Day para int, Month para int e Year para int
vendas_cafe['Transaction_ID'] = vendas_cafe['Transaction_ID'].astype('string')
vendas_cafe['Year'] = vendas_cafe['Year'].astype('Int64')
vendas_cafe['Month'] = vendas_cafe['Month'].astype('Int64')
vendas_cafe['Day'] = vendas_cafe['Day'].astype('Int64')
vendas_cafe['Week_Day'] = vendas_cafe['Week_Day'].astype('category')
```

```python
vendas_cafe.info()
```

```python
vendas_cafe.head()
```

```python
# Adicionar nova categoria na Localisação devido porcentagem alta de nulos mais de 30 por cento
# Categoria (Not Specified)
vendas_cafe['Location'] = (
    vendas_cafe['Location'].cat.add_categories('Not Specified').fillna('Not Specified')
)
```

```python
vendas_cafe.head()
```

```python
# Adicionar nova catgoria no Metodo_Pagamento devido porcentagem alta de nulos mais de 30 porcento
#Categoria (Not Specified)
vendas_cafe['Payment_Method']= (
    vendas_cafe['Payment_Method'].cat.add_categories('Not Specified').fillna('Not Specified')
)
```

```python
vendas_cafe.head()
```

```python
# Tratar volores da coluna 'Quantity' preencher com 1(minimo de venda)
vendas_cafe['Quantity'] = vendas_cafe['Quantity'].fillna(1)
```

```python
#Coluna Preco Unitario: preencher com mediana
mediana_por_produto = vendas_cafe.groupby('Item')['Price_Per_Unit'].transform('median')
vendas_cafe['Price_Per_Unit'] = vendas_cafe['Price_Per_Unit'].fillna(mediana_por_produto)
```

```python
vendas_cafe
```

```python
# Garantindo o preencimento com mediana casa algum produto tenho todos os precos nulos
vendas_cafe['Price_Per_Unit'] = vendas_cafe['Price_Per_Unit'].fillna(vendas_cafe['Price_Per_Unit'].median())
```

```python
# Recalcular Gasto Total
vendas_cafe['Total_Spent'] = vendas_cafe['Quantity'] * vendas_cafe['Price_Per_Unit']
```

```python
vendas_cafe
```
